```python


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, applications
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping, ModelCheckpoint
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report, mean_squared_error
import cv2
import os
import json
from datetime import datetime
import time
import warnings
warnings.filterwarnings('ignore')

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

class FacialExpressionAnalyzer:
    """
    A modular class for facial expression recognition and valence/arousal estimation
    using multiple CNN architectures.
    """
    
    def __init__(self, config):
        """
        Initialize the analyzer with configuration parameters.
        """
        self.config = config
        self.model = None
        self.history = None
        self.training_time = None
        
        # Create directories for saving results
        os.makedirs('models', exist_ok=True)
        os.makedirs('results', exist_ok=True)
        os.makedirs('plots', exist_ok=True)
        os.makedirs('qualitative_results', exist_ok=True)
    
    def load_and_preprocess_data(self, dataset_path='Dataset/Dataset/'):
        
        print("Loading and preprocessing REAL dataset...")
        
        try:
            self._load_real_dataset(dataset_path)
            print("Real dataset loaded successfully!")
            
        except Exception as e:
            print(f"Error loading real dataset: {e}")
            import traceback
            traceback.print_exc()
            print("Falling back to synthetic data...")
            self._create_synthetic_data()
    
    def _load_real_dataset(self, dataset_path):
        
        print("Loading real dataset...")
        
        annotations_path = os.path.join(dataset_path, 'annotations')
        images_path = os.path.join(dataset_path, 'images')
        
        print(f"Annotations path: {annotations_path}")
        print(f"Images path: {images_path}")
        
        # Get all annotation files
        annotation_files = os.listdir(annotations_path)
        print(f"Found {len(annotation_files)} annotation files")
        
        # Group files by their prefix (number before underscore)
        file_groups = {}
        for file in annotation_files:
            if '_' in file:
                prefix = file.split('_')[0]
                if prefix not in file_groups:
                    file_groups[prefix] = {}
                file_groups[prefix][file.split('_')[1].split('.')[0]] = file
        
        print(f"Found {len(file_groups)} file groups")
        
        # Load images
        image_files = [f for f in os.listdir(images_path) 
                      if f.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp'))]
        image_files.sort(key=lambda x: int(x.split('.')[0]))
        print(f"Found {len(image_files)} image files")
        
        # Create arrays to store data
        images = []
        expressions = []
        valence_values = []
        arousal_values = []
        
        img_size = self.config['image_size']
        
        # Process each file group
        for prefix, files in list(file_groups.items())[:1000]:  # Limit for testing
            try:
                # Check if we have all required files
                if 'exp' in files and 'val' in files and 'aro' in files:
                    # Load annotations
                    exp_data = np.load(os.path.join(annotations_path, files['exp']), allow_pickle=True)
                    val_data = np.load(os.path.join(annotations_path, files['val']), allow_pickle=True)
                    aro_data = np.load(os.path.join(annotations_path, files['aro']), allow_pickle=True)
                    
                    # Extract values safely
                    def extract_value(data):
                        if hasattr(data, 'shape'):
                            if data.shape == ():  # Scalar
                                return data.item()
                            else:  # Array
                                return data[0] if len(data) > 0 else 0
                        else:
                            return data
                    
                    expr = int(extract_value(exp_data))
                    val = float(extract_value(val_data))
                    aro = float(extract_value(aro_data))
                    
                    # Skip invalid samples
                    if val == -2 or aro == -2:
                        continue
                    
                    # Load corresponding image
                    image_file = f"{prefix}.jpg"
                    image_path = os.path.join(images_path, image_file)
                    
                    if os.path.exists(image_path):
                        img = cv2.imread(image_path)
                        if img is not None:
                            # Convert BGR to RGB
                            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                            
                            # Resize image
                            img = cv2.resize(img, img_size)
                            
                            # Normalize to [0, 1]
                            img = img.astype(np.float32) / 255.0
                            
                            images.append(img)
                            expressions.append(expr)
                            valence_values.append(val)
                            arousal_values.append(aro)
                            
            except Exception as e:
                continue
        
        if len(images) == 0:
            raise ValueError("No valid images were loaded from the real dataset")
        
        print(f"Successfully loaded {len(images)} real images with annotations")
        
        # Convert to numpy arrays
        self.X = np.array(images)
        self.y_expr = np.array(expressions)
        self.y_valence = np.array(valence_values)
        self.y_arousal = np.array(arousal_values)
        
        print(f"Dataset shape: {self.X.shape}")
        print(f"Expression distribution: {np.bincount(self.y_expr)}")
        
        # Split the dataset
        self._split_dataset()
    
    def _split_dataset(self):
        """
        Split the dataset into training, validation, and test sets.
        """
        # First split: separate test set
        X_temp, self.X_test, y_expr_temp, self.y_test_expr, \
        y_valence_temp, self.y_test_valence, y_arousal_temp, self.y_test_arousal = train_test_split(
            self.X, self.y_expr, self.y_valence, self.y_arousal,
            test_size=0.15, random_state=42, stratify=self.y_expr
        )
        
        # Second split: training and validation
        self.X_train, self.X_val, self.y_train_expr, self.y_val_expr, \
        self.y_train_valence, self.y_val_valence, self.y_train_arousal, self.y_val_arousal = train_test_split(
            X_temp, y_expr_temp, y_valence_temp, y_arousal_temp,
            test_size=0.15, random_state=42, stratify=y_expr_temp
        )
        
        print(f"Training set: {self.X_train.shape[0]} samples")
        print(f"Validation set: {self.X_val.shape[0]} samples")
        print(f"Test set: {self.X_test.shape[0]} samples")
        print(f"Final expression distribution - Train: {np.bincount(self.y_train_expr)}")
    
    def _create_synthetic_data(self):
        
        print("Creating synthetic data as fallback...")
        num_samples = 1000
        img_height, img_width = self.config['image_size']
        
        self.X_train = np.random.normal(0.5, 0.2, (int(num_samples * 0.7), img_height, img_width, 3)).astype(np.float32)
        self.X_val = np.random.normal(0.5, 0.2, (int(num_samples * 0.15), img_height, img_width, 3)).astype(np.float32)
        self.X_test = np.random.normal(0.5, 0.2, (int(num_samples * 0.15), img_height, img_width, 3)).astype(np.float32)
        
        self.X_train = np.clip(self.X_train, 0, 1)
        self.X_val = np.clip(self.X_val, 0, 1)
        self.X_test = np.clip(self.X_test, 0, 1)
        
        self.y_train_expr = np.random.choice(8, int(num_samples * 0.7), p=[0.2, 0.15, 0.1, 0.1, 0.1, 0.1, 0.15, 0.1])
        self.y_val_expr = np.random.choice(8, int(num_samples * 0.15), p=[0.2, 0.15, 0.1, 0.1, 0.1, 0.1, 0.15, 0.1])
        self.y_test_expr = np.random.choice(8, int(num_samples * 0.15), p=[0.2, 0.15, 0.1, 0.1, 0.1, 0.1, 0.15, 0.1])
        
        def create_affective_labels(expressions):
            valence = np.zeros(len(expressions))
            arousal = np.zeros(len(expressions))
            
            for i, expr in enumerate(expressions):
                if expr == 0:   valence[i], arousal[i] = 0.0, 0.0
                elif expr == 1: valence[i], arousal[i] = 0.7, 0.6
                elif expr == 2: valence[i], arousal[i] = -0.8, -0.4
                elif expr == 3: valence[i], arousal[i] = 0.4, 0.8
                elif expr == 4: valence[i], arousal[i] = -0.6, 0.7
                elif expr == 5: valence[i], arousal[i] = -0.7, 0.3
                elif expr == 6: valence[i], arousal[i] = -0.9, 0.8
                elif expr == 7: valence[i], arousal[i] = -0.5, -0.2
                
                valence[i] += np.random.normal(0, 0.1)
                arousal[i] += np.random.normal(0, 0.1)
            
            return np.clip(valence, -1, 1), np.clip(arousal, -1, 1)
        
        self.y_train_valence, self.y_train_arousal = create_affective_labels(self.y_train_expr)
        self.y_val_valence, self.y_val_arousal = create_affective_labels(self.y_val_expr)
        self.y_test_valence, self.y_test_arousal = create_affective_labels(self.y_test_expr)
        
        print("Synthetic data created as fallback")

    def create_data_augmentation(self):
        """Create data augmentation pipeline."""
        return keras.Sequential([
            layers.RandomFlip("horizontal"),
            layers.RandomRotation(0.1),
            layers.RandomZoom(0.1),
            layers.RandomContrast(0.1),
        ])
    
    def build_model(self, architecture='resnet50'):
        """Build a multi-task CNN model."""
        print(f"Building {architecture} model...")
        
        inputs = layers.Input(shape=(*self.config['image_size'], 3))
        x = self.create_data_augmentation()(inputs)
        
        if architecture == 'resnet50':
            base_model = applications.ResNet50(
                weights='imagenet', include_top=False, input_shape=(*self.config['image_size'], 3)
            )
        elif architecture == 'mobilenet':
            base_model = applications.MobileNetV2(
                weights='imagenet', include_top=False, input_shape=(*self.config['image_size'], 3)
            )
        elif architecture == 'custom':
            return self._build_custom_model(inputs)
        else:
            raise ValueError(f"Unsupported architecture: {architecture}")
        
        base_model.trainable = False
        
        x = base_model(x, training=False)
        x = layers.GlobalAveragePooling2D()(x)
        x = layers.Dense(512, activation='relu')(x)
        x = layers.Dropout(0.5)(x)
        x = layers.BatchNormalization()(x)
        x = layers.Dense(256, activation='relu')(x)
        x = layers.Dropout(0.3)(x)
        
        expression_output = layers.Dense(8, activation='softmax', name='expression')(x)
        valence_output = layers.Dense(1, activation='tanh', name='valence')(x)
        arousal_output = layers.Dense(1, activation='tanh', name='arousal')(x)
        
        model = models.Model(inputs=inputs, outputs=[expression_output, valence_output, arousal_output])
        self._compile_model(model)
        
        self.model = model
        self.architecture = architecture
        return model
    
    def _build_custom_model(self, inputs):
        x = layers.Conv2D(32, (3, 3), activation='relu')(inputs)
        x = layers.MaxPooling2D((2, 2))(x)
        x = layers.Conv2D(64, (3, 3), activation='relu')(x)
        x = layers.MaxPooling2D((2, 2))(x)
        x = layers.Conv2D(128, (3, 3), activation='relu')(x)
        x = layers.GlobalAveragePooling2D()(x)
        x = layers.Dense(128, activation='relu')(x)
        x = layers.Dropout(0.5)(x)
        
        expression_output = layers.Dense(8, activation='softmax', name='expression')(x)
        valence_output = layers.Dense(1, activation='tanh', name='valence')(x)
        arousal_output = layers.Dense(1, activation='tanh', name='arousal')(x)
        
        model = models.Model(inputs=inputs, outputs=[expression_output, valence_output, arousal_output])
        self._compile_model(model)
        
        self.model = model
        self.architecture = 'custom'
        return model
    
    def _compile_model(self, model):
        
        model.compile(
            optimizer=Adam(learning_rate=self.config['learning_rate']),
            loss={
                'expression': 'sparse_categorical_crossentropy',
                'valence': 'mse',
                'arousal': 'mse'
            },
            loss_weights={
                'expression': 1.0,
                'valence': 0.5,
                'arousal': 0.5
            },
            metrics={
                'expression': 'accuracy',
                'valence': 'mae',
                'arousal': 'mae'
            }
        )
    
    def train_model(self, X_train, y_train, X_val, y_val):
        
        print("Training model...")
        start_time = time.time()
        
        callbacks = [
            ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=3, min_lr=1e-7, verbose=1),
            EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True, verbose=1),
            ModelCheckpoint(f'models/best_{self.architecture}_model.h5', monitor='val_loss', save_best_only=True)
        ]
        
        self.history = self.model.fit(
            X_train, 
            {'expression': y_train[0], 'valence': y_train[1], 'arousal': y_train[2]},
            batch_size=self.config['batch_size'], 
            epochs=self.config['epochs'],
            validation_data=(X_val, {'expression': y_val[0], 'valence': y_val[1], 'arousal': y_val[2]}),
            callbacks=callbacks, 
            verbose=1
        )
        
        self.training_time = time.time() - start_time
        print(f"Training completed in {self.training_time:.2f} seconds")
    
    def evaluate_model(self, X_test, y_test):
        
        print("Evaluating model...")
        
        predictions = self.model.predict(X_test, verbose=0)
        expr_pred = np.argmax(predictions[0], axis=1)
        valence_pred = predictions[1].flatten()
        arousal_pred = predictions[2].flatten()
        
        metrics = self._calculate_metrics(y_test, expr_pred, valence_pred, arousal_pred)
        return metrics, (expr_pred, valence_pred, arousal_pred)
    
    def _calculate_metrics(self, y_test, expr_pred, valence_pred, arousal_pred):
        
        expr_accuracy = np.mean(expr_pred == y_test[0])
        expr_cm = confusion_matrix(y_test[0], expr_pred)
        
        valence_rmse = np.sqrt(mean_squared_error(y_test[1], valence_pred))
        arousal_rmse = np.sqrt(mean_squared_error(y_test[2], arousal_pred))
        
        try:
            valence_corr = np.corrcoef(y_test[1], valence_pred)[0, 1]
            arousal_corr = np.corrcoef(y_test[2], arousal_pred)[0, 1]
        except:
            valence_corr = 0
            arousal_corr = 0
        
        valence_sagr = np.mean(np.sign(y_test[1]) == np.sign(valence_pred))
        arousal_sagr = np.mean(np.sign(y_test[2]) == np.sign(arousal_pred))
        
        valence_ccc = self._concordance_corr(y_test[1], valence_pred)
        arousal_ccc = self._concordance_corr(y_test[2], arousal_pred)
        
        return {
            'expression_accuracy': expr_accuracy, 'expression_cm': expr_cm,
            'valence_rmse': valence_rmse, 'arousal_rmse': arousal_rmse,
            'valence_corr': valence_corr, 'arousal_corr': arousal_corr,
            'valence_sagr': valence_sagr, 'arousal_sagr': arousal_sagr,
            'valence_ccc': valence_ccc, 'arousal_ccc': arousal_ccc,
            'training_time': self.training_time
        }
    
    def _concordance_corr(self, y_true, y_pred):
        
        mask = ~(np.isnan(y_true) | np.isnan(y_pred))
        y_true, y_pred = y_true[mask], y_pred[mask]
        
        if len(y_true) < 2:
            return 0
            
        corr = np.corrcoef(y_true, y_pred)[0, 1]
        mean_true, mean_pred = np.mean(y_true), np.mean(y_pred)
        var_true, var_pred = np.var(y_true), np.var(y_pred)
        
        numerator = 2 * corr * np.sqrt(var_true) * np.sqrt(var_pred)
        denominator = var_true + var_pred + (mean_true - mean_pred) ** 2
        
        return numerator / denominator if denominator != 0 else 0

    def plot_training_history(self):
        
        if self.history is None:
            print("No training history available.")
            return
            
        history = self.history.history
        
        fig, axes = plt.subplots(2, 3, figsize=(15, 10))
        
        # Loss
        axes[0, 0].plot(history['loss'], label='Training')
        axes[0, 0].plot(history['val_loss'], label='Validation')
        axes[0, 0].set_title('Total Loss')
        axes[0, 0].legend()
        
        # Expression accuracy
        axes[0, 1].plot(history['expression_accuracy'], label='Training')
        axes[0, 1].plot(history['val_expression_accuracy'], label='Validation')
        axes[0, 1].set_title('Expression Accuracy')
        axes[0, 1].legend()
        
        # Expression loss
        axes[0, 2].plot(history['expression_loss'], label='Training')
        axes[0, 2].plot(history['val_expression_loss'], label='Validation')
        axes[0, 2].set_title('Expression Loss')
        axes[0, 2].legend()
        
        # Valence MAE
        axes[1, 0].plot(history['valence_mae'], label='Training')
        axes[1, 0].plot(history['val_valence_mae'], label='Validation')
        axes[1, 0].set_title('Valence MAE')
        axes[1, 0].legend()
        
        # Arousal MAE
        axes[1, 1].plot(history['arousal_mae'], label='Training')
        axes[1, 1].plot(history['val_arousal_mae'], label='Validation')
        axes[1, 1].set_title('Arousal MAE')
        axes[1, 1].legend()
        
        axes[1, 2].axis('off')
        
        plt.suptitle(f'Training History - {self.architecture}', fontsize=16)
        plt.tight_layout()
        plt.savefig(f'plots/{self.architecture}_training_history.png', dpi=300, bbox_inches='tight')
        plt.show()
    
    def plot_confusion_matrix(self, cm, class_names):
        
        plt.figure(figsize=(10, 8))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                   xticklabels=class_names, yticklabels=class_names)
        plt.title(f'Confusion Matrix - {self.architecture}')
        plt.xlabel('Predicted Label')
        plt.ylabel('True Label')
        plt.savefig(f'plots/{self.architecture}_confusion_matrix.png', dpi=300, bbox_inches='tight')
        plt.show()
    
    def plot_qualitative_results(self, X_test, y_test, predictions, class_names, num_samples=5):
        
        expr_pred, valence_pred, arousal_pred = predictions
        expr_true = y_test[0]
        
        correct_mask = expr_pred == expr_true
        incorrect_mask = expr_pred != expr_true
        
        correct_indices = np.where(correct_mask)[0][:num_samples]
        incorrect_indices = np.where(incorrect_mask)[0][:num_samples]
        
        self._plot_samples(X_test, expr_true, expr_pred, correct_indices, class_names, "Correct")
        self._plot_samples(X_test, expr_true, expr_pred, incorrect_indices, class_names, "Incorrect")
    
    def _plot_samples(self, X, y_true, y_pred, indices, class_names, title):
        
        if len(indices) == 0:
            return
            
        n_samples = min(len(indices), 5)
        fig, axes = plt.subplots(1, n_samples, figsize=(n_samples * 3, 3))
        
        if n_samples == 1:
            axes = [axes]
        
        for i, idx in enumerate(indices[:n_samples]):
            axes[i].imshow(X[idx])
            axes[i].set_title(f'True: {class_names[y_true[idx]]}\nPred: {class_names[y_pred[idx]]}')
            axes[i].axis('off')
        
        plt.suptitle(f'{title} Predictions - {self.architecture}', fontsize=14)
        plt.tight_layout()
        plt.savefig(f'qualitative_results/{self.architecture}_{title.lower()}.png', dpi=300)
        plt.show()
    
    def generate_report(self, metrics, architecture):
        
        report = {
            'architecture': architecture,
            'timestamp': datetime.now().isoformat(),
            'parameters': self.config,
            'metrics': metrics
        }
        
        with open(f'results/{architecture}_report.json', 'w') as f:
            json.dump(report, f, indent=4)
        
        print(f"Report saved for {architecture}")

def main():
    
    # Configuration
    config = {
        'batch_size': 32,
        'epochs': 20,
        'learning_rate': 0.001,
        'image_size': (224, 224),
        'architectures': ['resnet50', 'mobilenet', 'custom']
    }
    
    expression_classes = ['Neutral', 'Happy', 'Sad', 'Surprise', 'Fear', 'Disgust', 'Anger', 'Contempt']
    
    # Initialize analyzer
    analyzer = FacialExpressionAnalyzer(config)
    
    # Load REAL dataset
    analyzer.load_and_preprocess_data(dataset_path='Dataset/Dataset/')
    
    # Prepare data
    X_train, X_val, X_test = analyzer.X_train, analyzer.X_val, analyzer.X_test
    y_train = (analyzer.y_train_expr, analyzer.y_train_valence, analyzer.y_train_arousal)
    y_val = (analyzer.y_val_expr, analyzer.y_val_valence, analyzer.y_val_arousal)
    y_test = (analyzer.y_test_expr, analyzer.y_test_valence, analyzer.y_test_arousal)
    
    print(f"Real dataset loaded successfully!")
    print(f"Training set: {X_train.shape}")
    print(f"Validation set: {X_val.shape}")
    print(f"Test set: {X_test.shape}")
    
    # Train and evaluate architectures
    results_comparison = {}
    
    for arch in config['architectures']:
        print(f"\n{'='*60}")
        print(f"Training {arch} architecture on REAL DATA")
        print(f"{'='*60}")
        
        try:
            # Build and train model
            model = analyzer.build_model(architecture=arch)
            
            if arch == 'resnet50':
                print("Model summary:")
                model.summary()
            
            analyzer.train_model(X_train, y_train, X_val, y_val)
            
            # Generate plots
            analyzer.plot_training_history()
            metrics, predictions = analyzer.evaluate_model(X_test, y_test)
            analyzer.plot_confusion_matrix(metrics['expression_cm'], expression_classes)
            analyzer.plot_qualitative_results(X_test, y_test, predictions, expression_classes)
            
            # Save results
            analyzer.generate_report(metrics, arch)
            results_comparison[arch] = metrics
            
            print(f"\n{arch} Results:")
            print(f"  Expression Accuracy: {metrics['expression_accuracy']:.4f}")
            print(f"  Valence RMSE: {metrics['valence_rmse']:.4f}")
            print(f"  Arousal RMSE: {metrics['arousal_rmse']:.4f}")
            print(f"  Training Time: {metrics['training_time']:.2f}s")
            
        except Exception as e:
            print(f"Error with {arch}: {e}")
            import traceback
            traceback.print_exc()
        
        tf.keras.backend.clear_session()
    
    # Generate comparison report
    print(f"\n{'='*60}")
    print("FINAL ARCHITECTURE COMPARISON (REAL DATASET)")
    print(f"{'='*60}")
    
    for arch, metrics in results_comparison.items():
        print(f"{arch}:")
        print(f"  Accuracy: {metrics['expression_accuracy']:.4f}")
        print(f"  Valence RMSE: {metrics['valence_rmse']:.4f}")
        print(f"  Arousal RMSE: {metrics['arousal_rmse']:.4f}")
        print(f"  Valence CCC: {metrics['valence_ccc']:.4f}")
        print(f"  Arousal CCC: {metrics['arousal_ccc']:.4f}")
        print(f"  Time: {metrics['training_time']:.2f}s")
        print()

if __name__ == "__main__":
    main()
```
